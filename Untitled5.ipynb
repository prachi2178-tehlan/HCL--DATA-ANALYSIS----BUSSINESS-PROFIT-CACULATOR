```python
"""
PROJECT 15 : Bussiness Profit Calculator
"""

def get_positive_number(prompt, allow_zero=False):
    """Read a valid non-negative number (or positive number) from the user."""
    while True:
        try:
            value = float(input(prompt))
            if value < 0 or (not allow_zero and value == 0):
                print("Please enter a valid number.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a number.")


def get_positive_integer(prompt):
    """Read a valid positive whole number from the user."""
    while True:
        try:
            value = int(input(prompt))
            if value <= 0:
                print("Please enter a whole number greater than zero.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a whole number.")


def main():
    print("=" * 48)
    print("          SMART BUSINESS PROFIT CALCULATOR")
    print("=" * 48)

    business_name = input("Enter business name: ").strip() or "Not specified"
    product_name = input("Enter product name: ").strip() or "Not specified"

    quantity = get_positive_integer("Enter units sold: ")
    cost_price = get_positive_number("Enter cost price per unit (Rs.): ", allow_zero=True)
    selling_price = get_positive_number("Enter selling price per unit (Rs.): ", allow_zero=True)
    expenses = get_positive_number("Enter additional expenses (Rs.): ", allow_zero=True)

    # Business calculations
    total_revenue = selling_price * quantity
    product_cost = cost_price * quantity
    total_cost = product_cost + expenses
    net_profit = total_revenue - total_cost

    if total_revenue > 0:
        profit_margin = (net_profit / total_revenue) * 100
    else:
        profit_margin = 0.0

    # Display report
    print("\n" + "=" * 48)
    print("                 BUSINESS REPORT")
    print("=" * 48)
    print(f"Business name       : {business_name}")
    print(f"Product name        : {product_name}")
    print(f"Units sold          : {quantity}")
    print("-" * 48)
    print(f"Total revenue       : Rs. {total_revenue:,.2f}")
    print(f"Product cost        : Rs. {product_cost:,.2f}")
    print(f"Additional expenses : Rs. {expenses:,.2f}")
    print(f"Total cost          : Rs. {total_cost:,.2f}")
    print("-" * 48)
    print(f"Net profit/loss     : Rs. {net_profit:,.2f}")
    print(f"Profit margin       : {profit_margin:.2f}%")

    if net_profit > 0:
        print("Business status     : PROFIT")
    elif net_profit < 0:
        print("Business status     : LOSS")
    else:
        print("Business status     : BREAK-EVEN")

    print("=" * 48)
    print("Thank you for using Smart Business Profit Calculator!")


if __name__ == "__main__":
    main()
```
